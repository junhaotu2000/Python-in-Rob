### 2.1 Dataset类

```python
'''
    ├── train
    │   ├── ants
    │   └── bees
    └── val
        ├── ants
        └── bees
    '''
class MyData_folder(Dataset):
    '''
    从Dataset里继承创建适用于classification任务的数据预处理类
    ''' 
    # 读取 data_path, label_path
    def __init__(self, root_dir, label):
        self.root_dir = root_dir
        self.label = label
        self.img_dir = os.path.join(self.root_dir, self.label)
        self.img_name_list = os.listdir(self.img_dir)

    # 返回 data, label
    def __getitem__(self, idx):
        img_name = self.img_name_list[idx]
        img_path = os.path.join(self.root_dir, self.label, img_name)
        img_PIL = Image.open(img_path)
        label = self.label
        return img_PIL, label

    # 返回 一类数据的大小
    def __len__(self):
        return len(self.img_name_list)
```

```python
from torch.utils.data import Dataset
from PIL import Image

class MyData(Dataset):
    def __init__(self, img_path: list, img_class: list, transforms):
        self.img_path = img_path
        self.img_class = img_class
        self.transforms = transforms

    def __len__(self):
        return len(self.img_path)

    def __getitem__(self, idx):
        img = self.transforms(Image.open(self.img_path[idx]))
        label = self.image_class[idx]

        return img, label
```

```python
'''
    └── ├── train
    │   ├── ants_image
    │   ├── ants_label
    │   ├── bees_image
    │   └── bees_label
    └── val
        ├── ants_image
        ├── ants_label
        ├── bees_image
        └── bees_label
    '''

class MyData_label(Dataset):
    '''
    从Dataset里继承创建适用于object indification任务的数据预处理类
    ''' 
    def __init__(self, img_dir, label_dir):
        self.img_dir = img_dir
        self.label_dir = label_dir
        self.img_name_list = os.listdir(self.img_dir)
        self.label_file_list= os.listdir(self.label_dir)

    def __getitem__(self, idx):
        img_name = self.img_name_list[idx]
        img_path = os.path.join(self.img_dir, img_name)
        img_data = Image.open(img_path)

        label_file_name = self.label_file_list[idx]
        label_file_path = os.path.join(self.label_dir, label_file_name)
        with open(label_file_path, 'r') as f:
            label = f.read()

        return img_data, label

    def __len__(self):
        return len(self.img_name_list)
```

将文件格式从folder改为但张图片label:

```python
def rename_img_num(img_path: str)-> bool:
    img_name_list = os.listdir(img_path)
    try:
        for i, img_name in enumerate(img_name_list):
            # get old path of imgs
            img_old_path = os.path.join(img_path, img_name)
            
            # get new path of imgs
            new_img_name = f'{i}{os.path.splitext(img_name)[1]}'
            img_new_path = os.path.join(img_path, new_img_name)
    
            # rename old path with new path
            os.rename(img_old_path, img_new_path)
        print('Rename successfully')
    except: 
        print('Rename failed')

def create_label_dir(img_path: str, des_name: str, label:str)-> bool:
    # make a new label folder
    parent_dir = os.path.dirname(img_path)
    des_path = os.path.join(parent_dir, des_name)
    if not os.path.exists(des_path):
        os.makedirs(des_path)
        
    # create label file '.txt'
    num_of_label = len(os.listdir(img_path))
    for i in range(num_of_label):
        file_name = f"{i}.txt"  
        file_path = os.path.join(des_path, file_name)
        with open(file_path, 'w') as file:
            file.write(label)
            
    print('Add label successfully')
```

## 4. 图像预处理 - Transforms的使用

`torchvision.transforms` 库是 PyTorch 中专门用于图像预处理的模块。它提供了**一系列常用的图像转换操作**，这些操作可以方便地对 `PIL` 图像、`NumPy` 数组或 PyTorch 张量进行处理。`transforms` 库中的操作主要用于在深度学习任务中对数据进行**预处理和增强**，比如图像的缩放、裁剪、归一化等。

```python
from torchvision import transforms
from PIL import Image
import cv2
```

使用流程：从`transforms`库中取出原子功能 -> 构建具体的工具 -> 将`opencv`或者`PIL`读取的图片先转换为`tensor`再放进工具里 -> 输出

使用技巧： 关注输入和输出, 关注参数

```python
# 取出原子功能 -> 构建具体的工具
'''convert ndarray or cv2 into tensor'''
img_to_tensor = transforms.ToTensor()
# 读取的图片
img_PIL = Image.open(f'dataset_mod/train/bees_image/1.jpg')
img_cv2 = cv2.imread(f'dataset_mod/train/bees_image/1.jpg')
# ToTensor (将opencv或者PIL读取的图片放进工具里 -> 输出)
img_tensor1 = img_to_tensor(img_PIL)
img_tensor2 =  img_to_tensor(img_cv2)
# tensor包含用于学习的大量参数
writer = SummaryWriter('imgs')
writer.add_image('Tensor_img1', img_tensor1) # tensorboard最好传入tensor数据类型
writer.add_image('Tensor_img2', img_tensor2)

# Normalize 
trans_norm = transforms.Normalize([0.05, 0.5, 0.5], [0.5, 0.5, 0.5]) # first_p: mean, second_p = std
img_norm = trans_norm(img_tensor1)
writer.add_image('Norm_img', img_norm)

# Resize 
trans_resize = transforms.Resize((512, 512)) # transforms.Resize(512)
img_resize = trans_resize(img_tensor1)
writer.add_image('Resize_img', img_resize)

# Compose 
trans_compose = transforms.Compose([trans_norm, trans_resize, img_to_tensor]) # 输入list包含transforms对象
img_compose = trans_compose(img_PIL)
writer.add_image('Compose_img', img_compose)

# RandomCrop
trans_random = transforms.RandomCrop(100)
trans_compose2 = transforms.Compose([trans_random, img_to_tensor]) # 先crop在tensor
for i in range(100):
    img_crop = trans_compose2(img_PIL)
    writer.add_image('RandCrop_img', img_crop, i)
```

## 7. 神经网络

```python
import torch.nn as nn
import torch.nn.functional as F
```

- `torch.nn`用于在继承的新模型的`__init__`
- `torch.nn.functional`用于在继承的新模型的`forward`函数中定义激活函数h
- 二者在设计上的区别：其实主要的原因在乎我们构建计算图的时候，有些操作不需要进行体现在计算图中的，例如ReLu层，池化层。但是像卷积层、全连接层还是需要体现在计算图中的。如果所有的层我们都用torch.nn.functional来定义，那么我们需要将卷积层和全连接层中的weights、bias全部手动写入计算图中去，这样是非常不方便的。如果我们全部使用类的方式来构建计算图，这样即使是非常简单的操作都需要构建类，这样是写代码的效率是非常低的。所以我们将卷积层、全连接层使用类的方式来进行定义，将池化和激活操作使用函数的方式进行使用，这样使我们更方便的构建计算图。

### 7.1 神经网络的基本骨架 - nn.Module

![image](./Media/conv2d.png)

```python
# 继承成nn.module, 只需要再定义__init__和foward函数
class Model(nn.Module):
    # 用于初始化layer
    def __init__(self):
        super(Model, self).__init__()
        self.conv1 = nn.Conv2d(1, 20, 5)
        self.conv2 = nn.Conv2d(20, 20, 5)

    # 用激活函数将每层连起来
    def forward(self, x):
        x = F.relu(self.conv1(x))
        return F.relu(self.conv2(x))
```

### 7.2 神经网络 - 卷积层 

```python
class ConvolutionModel(nn.Module):
    # torch.nn.Conv2d(in_channel, out_channel, kernel_size, stride, padding)
    # The 2D convolution (conv2d) in PyTorch expects a 4D input tensor with the following shape:
    '''(N, C_in, H, W)
        N: Batch size (number of input samples in a batch)
        C_in: Number of input channels (e.g., 1 for grayscale images, 3 for RGB images)
        H: Height of the input (number of rows)
        W: Width of the input (number of columns)
    '''
    def __init__(self):
        super(ConvolutionModel, self).__init__()
        self.conv1 = nn.Conv2d(1, 4, kernel_size = 2, stride = 1, padding=0)

    def forward(self, input):
        output = self.conv1(input)
        return output


input = torch.tensor([[1, 2, 0, 3, 1],
                      [1, 2, 1, 3, 1],
                      [2, 2, 0, 2, 1],
                      [1, 2, 3, 3, 2],
                      [1, 3, 0, 2, 1]], dtype=torch.float32).reshape(1, 1, 5, 5)
conv = ConvolutionModel()
conv(input)
```

```python
# torch.nn.functional.conv2d(input_matrix, kernel_matrix, stride, padding)

input = torch.tensor([[1, 2, 0, 3, 1],
                      [1, 2, 1, 3, 1],
                      [2, 2, 0, 2, 1],
                      [1, 2, 3, 3, 2],
                      [1, 3, 0, 2, 1]]).reshape(1, 1, 5, 5)

kernel = torch.tensor([[1, 2, 1],
                       [0, 1, 0],
                       [2, 1, 0]]).reshape(1, 1, 3, 3)

output = F.conv2d(input, kernel, stride=1)
print(output)

output1 = F.conv2d(input, kernel, stride=2)
print(output1)

output2 = F.conv2d(input, kernel, stride=1, padding=1)
print(output2)
```

### 7.3 神经网络 - 最大池化

[](Pytorch/Media/ceil_mode.png)

![image](./Media/ceil_mode.png)
![image](./Media/conv2d.png)

```python
import torch
from torch import nn

class MaxPool(nn.Module):
    '''
    ceil_mode: 决定在无法满足卷积核尺寸的情况下是否保留 (True： 保留 / False: 不保留)
    '''
    def __init__(self):
        super(MaxPool, self).__init__()
        # torch.nn.maxpool2d(kernel_zize, stride, padding, dilation, ceil_mode), ceil_mode默认为False
        self.maxpool = nn.MaxPool2d(kernel_size=3, ceil_mode=True)

    def forward(self, input):
        output = self.maxpool(input)
        return output
        
input = torch.tensor([[1, 2, 0, 3, 1],
                      [1, 2, 1, 3, 1],
                      [2, 2, 0, 2, 1],
                      [1, 2, 3, 3, 2],
                      [1, 3, 0, 2, 1]], dtype=torch.float32).reshape(1, 1, 5, 5)

kernel = torch.tensor([[1, 2, 1],
                       [0, 1, 0],
                       [2, 1, 0]]).reshape(1, 1, 3, 3)

pool = MaxPool()
pool(input)
```

```python
dataset = torchvision.datasets.CIFAR10('./datasets', train = False, download = True, transform=torchvision.transforms.ToTensor())
dataloader = DataLoader(dataset=dataset, batch_size=64, shuffle=False, num_workers=0, drop_last=True)
writer = SummaryWriter('logs_maxpool')

dataloader = tqdm(dataloader)
for step, data in enumerate(dataloader):
    imgs, labels = data
    writer.add_images("input", imgs, step)
    output = pool(imgs)
    writer.add_images("output", output, step)

    dataloader.desc = f'Step: {step}'
    dataloader.unit = 'step'
writer.close()
```

```python
class Nonlinear(nn.Module):

    def __init__(self):
        super(Nonlinear, self).__init__()
        self.nonlinear = nn.ReLU()

    def forward(self, input):
        output = self.nonlinear(input)
        return output

input = torch.tensor([[1, 2, 0, 3, 1],
                      [1, 2, 1, 3, 1],
                      [2, -2, 0, 2, 1],
                      [1, 2, 3, -3, 2],
                      [1, 3, 0, 2, 1]], dtype=torch.float32).reshape(1, 1, 5, 5)
nonlinear = Nonlinear()
print(nonlinear(input))

print(F.relu(input))
```

### 7.6 神经网络 - 实际搭建和Sequential的使用

![image](./Media/CIFAR10.png)

```python
class CIFAR10(nn.Module):
    '''
    使用Sequential连接网络可以避免在forward函数里逐一调用网络
    '''
    def __init__(self):
        super().__init__() # 记得要继承nn.Module
        self.model1 = nn.Sequential(
            nn.Conv2d(3, 32, 5, padding=2),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 32, 5, padding=2),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 5, padding=2),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(1024, 64),
            nn.Linear(64, 10),
            )

    def forward(self, input):
        output = self.model1(input)
        return output

# 网络错位第一也能成功初始化，需要额外测试性能
model = CIFAR10()
print(model) 

# 可以随机用一个sample，来测试网络的节点输出尺寸
sample = torch.ones((64, 3, 32, 32))
print(model(sample).shape)

# tensorboard也可以用来可视化模型
writer = SummaryWriter('./model')
writer.add_graph(model, sample)
writer.close()
```

```python
class CIFAR10(nn.Module):
    '''
    使用Sequential连接网络可以避免在forward函数里逐一调用网络
    '''
    def __init__(self):
        super().__init__() # 记得要继承nn.Module
        self.model1 = nn.Sequential(
            nn.Conv2d(3, 32, 5, padding=2),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 32, 5, padding=2),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 5, padding=2),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(1024, 64),
            nn.Linear(64, 10),
            )

    def forward(self, input):
        output = self.model1(input)
        return output
        
dataset = torchvision.datasets.CIFAR10('./datasets', train = False, download = True, transform=torchvision.transforms.ToTensor())
dataloader = DataLoader(dataset=dataset, batch_size=64, shuffle=False, num_workers=0, drop_last=True)
model = CIFAR10()
loss = nn.CrossEntropyLoss()

for data in dataloader:
    imgs, targets = data
    outputs = model(imgs)
    result_loss = loss(outputs, targets)
    result_loss.backward()
    print("OK")
```

### 7.8. 优化器

```python
from tqdm import tqdm  # Import tqdm properly

dataset = torchvision.datasets.CIFAR10('./datasets', train=False, download=True, transform=torchvision.transforms.ToTensor())
dataloader = DataLoader(dataset=dataset, batch_size=128, shuffle=True, num_workers=0, drop_last=True)

model = CIFAR10()
loss = nn.CrossEntropyLoss()
optim = torch.optim.SGD(model.parameters(), lr=0.01)

# Wrap the dataloader with tqdm for progress tracking
# dataloader = tqdm(dataloader)

for epoch in range(100):
    running_loss = 0.0
    for data in tqdm(dataloader):
        imgs, targets = data
        outputs = model(imgs)
        result_loss = loss(outputs, targets)
        
        optim.zero_grad()  # Zero the gradients
        result_loss.backward()  # Backpropagate the gradients
        optim.step()  # Update the parameters

        running_loss = running_loss + result_loss
        
     print(f"The loss of {epoch} is {running_loss}")
```

### 8.2 完整的模型训练套路

```python
# 准备数据集
train_data = torchvision.datasets.CIFAR10('./datasets', train=True, download=True, transform=torchvision.transforms.ToTensor())
test_data = torchvision.datasets.CIFAR10('./datasets', train=False, download=True, transform=torchvision.transforms.ToTensor())

# 数据长度
train_data_size = len(train_data)
test_data_size = len(test_data)
print(f"训练数据集的长度: {train_data_size}")
print(f"测试数据集的长度: {test_data_size}")

# 利用dataloader来加载数据集
train_dataloader = DataLoader(dataset = train_data, batch_size = 64)
test_dataloader = DataLoader(dataset = test_data, batch_size = 64)

# 搭建神经网络 (这个一般放在单独的model文件里)
class Cifar10(nn.Module):

    def __init__(self):
        super().__init__()
        self.Cifar10 = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=5, padding=2),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 32, kernel_size=5, padding=2),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, kernel_size=5, padding=2),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(1024, 64),
            nn.Linear(64, 10),
        )

    def forward(self, input):
        output = self.Cifar10(input)
        return output

# 创建网络模型
model = Cifar10()

# 创建损失函数
loss_cross = nn.CrossEntropyLoss()

# 定义优化器
optim = torch.optim.SGD(model.parameters(), lr=1e-2)

# 设置训练网络的参数
total_train_step = 0
# 记录测试的次数
total_test_step = 0
# 训练的轮数
epoch = 10

# 使用tensorboard可视化训练结果
writer = SummaryWriter('logs')

for i in range(epoch):
    ## 训练步骤开始
    model.train() # 对部分层有作用
    for iter, data in enumerate(tqdm(train_dataloader)):
        imgs, targets = data
        outputs = model(imgs)
        loss = loss_cross(outputs, targets)
        optim.zero_grad()
        loss.backward()
        optim.step()
        total_train_step += 1

        if total_test_step % 100 == 0:
            print(f"训练次数：{total_train_step}, Loss: {loss.item()}") # .item()将tensor转化为数字
            writer.add_scalar("train_loss", loss.item(), total_train_step)

    ## 测试步骤开始
    model.eval() # 对部分层有作用
    total_test_loss = 0
    with torch.no_grad():
        for data in test_dataloader:
            imgs, targets = data
            outputs = model(imgs)
            loss = loss_cross(outputs, targets)
            total_test_loss += loss
            accuracy = (output.argmax(1) == targets).sum()
            total_accuracy = total_accuracy + accuracy
    
    print(f'整体测试集上的Loss: {total_test_loss}')
    print(f'整体测试集上的Accuracy: {total_accuracy/test_data_size}')
    writer.add_scalar("test_loss", total_test_loss.item(), total_test_step)
    writer.add_scalar("test_accuracy", total_accuracy/test_data_size, total_test_step)
    total_tes_step += 1
    
    torch.save(model, f"tudui_{i}.pth")
    print("模型已保存")
writer.close()
```

### 8.3 GPU训练

![image](./Media/cuda.png)

```python
# 准备数据集
train_data = torchvision.datasets.CIFAR10('./datasets', train=True, download=True, transform=torchvision.transforms.ToTensor())
test_data = torchvision.datasets.CIFAR10('./datasets', train=False, download=True, transform=torchvision.transforms.ToTensor())

# 数据长度
train_data_size = len(train_data)
test_data_size = len(test_data)
print(f"训练数据集的长度: {train_data_size}")
print(f"测试数据集的长度: {test_data_size}")

# 利用dataloader来加载数据集
train_dataloader = DataLoader(dataset = train_data, batch_size = 64)
test_dataloader = DataLoader(dataset = test_data, batch_size = 64)

# 搭建神经网络 (这个一般放在单独的model文件里)
class Cifar10(nn.Module):

    def __init__(self):
        super().__init__()
        self.Cifar10 = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=5, padding=2),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 32, kernel_size=5, padding=2),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, kernel_size=5, padding=2),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(1024, 64),
            nn.Linear(64, 10),
        )

    def forward(self, input):
        output = self.Cifar10(input)
        return output

# 创建网络模型
model = Cifar10()
if torch.cuda.is_available():
    '''1. 将模型转化到GPU上'''
    model = model.cuda() 

# 创建损失函数
loss_cross = nn.CrossEntropyLoss()
if torch.cuda.is_available():
    '''2. 将损失函数转化到GPU上'''
    loss_cross = loss_cross.cuda() 

# 定义优化器
optim = torch.optim.SGD(model.parameters(), lr=1e-2)

# 设置训练网络的参数
total_train_step = 0
# 记录测试的次数
total_test_step = 0
# 训练的轮数
epoch = 10

# 使用tensorboard可视化训练结果
writer = SummaryWriter('logs')

for i in range(epoch):
    ## 训练步骤开始
    model.train() # 对部分层有作用
    for iter, data in enumerate(tqdm(train_dataloader)):
        imgs, targets = data
        if torch.cuda.is_available():
            '''3. 将数据集转换到GPU上'''
            imgs = imgs.cuda()
            targets = targets.cuda()
        outputs = model(imgs)
        loss = loss_cross(outputs, targets)
        optim.zero_grad()
        loss.backward()
        optim.step()
        total_train_step += 1

        if total_test_step % 100 == 0:
            print(f"训练次数：{total_train_step}, Loss: {loss.item()}") # .item()将tensor转化为数字
            writer.add_scalar("train_loss", loss.item(), total_train_step)

    ## 测试步骤开始
    model.eval() # 对部分层有作用
    total_test_loss = 0
    total_sccuracy = 0
    with torch.no_grad():
        for data in test_dataloader:
            imgs, targets = data
            if torch.cuda.is_available():
                '''3. 将数据集转换到GPU上'''
                imgs = imgs.cuda() 
                targets = targets.cuda()
            outputs = model(imgs)
            loss = loss_cross(outputs, targets)
            total_test_loss += loss
            accuracy = (outputs.argmax(1) == targets).sum()
            total_accuracy = total_sccuracy + accuracy
    
    print(f'整体测试集上的Loss: {total_test_loss}')
    print(f'整体测试集上的Accuracy: {total_sccuracy/test_data_size}')
    writer.add_scalar("test_loss", total_test_loss.item(), total_test_step)
    writer.add_scalar("test_accuracy", total_sccuracy/test_data_size, total_test_step)
    total_test_step += 1
    
    torch.save(model, f"tudui_{i}.pth")
    print("模型已保存")
writer.close()
```

```python
# 定义训练的设备
device = torch.device("cuda:0")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)

# 准备数据集
train_data = torchvision.datasets.CIFAR10('./datasets', train=True, download=True, transform=torchvision.transforms.ToTensor())
test_data = torchvision.datasets.CIFAR10('./datasets', train=False, download=True, transform=torchvision.transforms.ToTensor())

# 数据长度
train_data_size = len(train_data)
test_data_size = len(test_data)
print(f"训练数据集的长度: {train_data_size}")
print(f"测试数据集的长度: {test_data_size}")

# 利用dataloader来加载数据集
train_dataloader = DataLoader(dataset = train_data, batch_size = 64)
test_dataloader = DataLoader(dataset = test_data, batch_size = 64)

# 搭建神经网络 (这个一般放在单独的model文件里)
class Cifar10(nn.Module):

    def __init__(self):
        super().__init__()
        self.Cifar10 = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=5, padding=2),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 32, kernel_size=5, padding=2),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, kernel_size=5, padding=2),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(1024, 64),
            nn.Linear(64, 10),
        )

    def forward(self, input):
        output = self.Cifar10(input)
        return output

# 创建网络模型
'''1. 将模型转化到GPU上'''
model = Cifar10().to(device)

# 创建损失函数
'''2. 将损失函数转化到GPU上'''
loss_cross = nn.CrossEntropyLoss().to(device) 

# 定义优化器
optim = torch.optim.SGD(model.parameters(), lr=1e-2)

# 设置训练网络的参数
total_train_step = 0
# 记录测试的次数
total_test_step = 0
# 训练的轮数
epoch = 10

# 使用tensorboard可视化训练结果
writer = SummaryWriter('logs')

for i in range(epoch):
    ## 训练步骤开始
    model.train() # 对部分层有作用
    for iter, data in enumerate(tqdm(train_dataloader)):
        imgs, targets = data
        '''3. 将数据集转换到GPU上'''
        imgs = imgs.to(device)
        targets = targets.to(device)
        outputs = model(imgs)
        loss = loss_cross(outputs, targets)
        optim.zero_grad()
        loss.backward()
        optim.step()
        total_train_step += 1

        if total_test_step % 100 == 0:
            print(f"训练次数：{total_train_step}, Loss: {loss.item()}") # .item()将tensor转化为数字
            writer.add_scalar("train_loss", loss.item(), total_train_step)

    ## 测试步骤开始
    model.eval() # 对部分层有作用
    total_test_loss = 0
    total_sccuracy = 0
    with torch.no_grad():
        for data in test_dataloader:
            imgs, targets = data
            '''3. 将数据集转换到GPU上'''
            imgs = imgs.to(device)
            targets = targets.to(device)
            outputs = model(imgs)
            loss = loss_cross(outputs, targets)
            total_test_loss += loss
            accuracy = (outputs.argmax(1) == targets).sum()
            total_accuracy = total_sccuracy + accuracy
    
    print(f'整体测试集上的Loss: {total_test_loss}')
    print(f'整体测试集上的Accuracy: {total_sccuracy/test_data_size}')
    writer.add_scalar("test_loss", total_test_loss.item(), total_test_step)
    writer.add_scalar("test_accuracy", total_sccuracy/test_data_size, total_test_step)
    total_test_step += 1
    
    torch.save(model, f"tudui_{i}.pth")
    print("模型已保存")
writer.close()
```
