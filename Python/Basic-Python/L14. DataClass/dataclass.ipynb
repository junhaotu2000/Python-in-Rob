## Dataclass and Typing
- [dataclass](https://www.dataquest.io/blog/how-to-use-python-data-classes/)

```python
from dataclasses import dataclass, field
from typing import Tuple, List, Union, Optional
```

### Define a class using dataclasses (`__init__`)

```python
@dataclass
class Person():
    house_coordinates: Tuple
    name: str = 'Joe'
    age: int =30
    height: float = 1.85
    email: str = 'joe@umich.edu'

@dataclass
class People():
    person: List[Person]

person1 = Person( (40.748441, -73.985664), 'Joe', 25, 1.85, 'joe@umich.edu')
print(person1)
    
person2 = Person( (40.748441, -73.985664), 'Mary', 25, 1.85, 'joe@umich.edu')
print(person2)

people = People([person1, person2])
print(people)
```

### Compare data classes (`__eq__`)

```python
@dataclass
class Person():
    '''auto write __eq__ function for you'''
    name: str = 'Joe'
    age: int = 30
    height: float = 1.85
    email: str = 'joe@dataquest.io'

    # TODO:overwrite the default method of equalization
    def __eq__(self, other):
        if isinstance(other, Person):
            return (self.name, self.age,
                    self.height, self.email) == (other.name, other.age,
                                                 other.height, other.email)
        return NotImplemented

print(Person() == Person())
```

### Sort data classes排序数据类

```python
@dataclass(order=True)
class Person():
    sort_index: int = field(init = False, repr = False)
    name: str
    age: int
    height: float
    email: str
    
    # 指定排序的对象
    def __post_init__(self):
        self.sort_index = self.age

joe = Person('Joe', 25, 1.85, 'joe@dataquest.io')
mary = Person('Mary', 43, 1.67, 'mary@dataquest.io')

print(joe > mary)
```

### Work with immutable data classes

```python
@dataclass(frozen=True)
class Person():
    name: str 
    age: int
    address: str
    email: str
    
person1 = Person(name='Joy', age = 16, address = 'NA', email='joy@umich.edu')
person1.age = 18
```

```python
@dataclass(frozen=True)
class People():
    people: List[Person]

@dataclass(frozen=True)
class Person():
    name: str
    age: int
    height: float
    email: str
    
joe = Person('Joe', 25, 1.85, 'joe@dataquest.io')
mary = Person('Mary', 43, 1.67, 'mary@dataquest.io')
two_people = People([joe, mary])
print(two_people)

# 不直接修改参数，而是替换参数的内部的mutable参数是可以的
# 因此frozen的参数内部也需要是frozen的
two_people.people[0] = Person('Joe', 35, 1.85, 'joe@dataquest.io')
print(two_people.people[0]) 
```

### Use inheritance with data classes

```python
@dataclass(order=True)
class Person():
    name: str
    age: int
    height: float
    email: str
    
@dataclass(order=True)
class Employee(Person):
    salary: int
    department: str
    
print(Employee('Joe', 25, 1.85, 'joe@dataquest.io', 100000, 'Marketing'))
```

```python
@dataclass
class Person():
    name: str = 'Joe'
    age: int = 30
    height: float = 1.85
    email: str = 'joe@dataquest.io'

@dataclass(order=True)
class Employee(Person):
    salary: int
    departament: str

# base class里有default value, 使用child base的时候就不能
print(Employee('Joe', 25, 1.85, 'joe@dataquest.io', 100000, 'Marketing'))
```

### Example

```python
from dataclasses import dataclass, field
from typing import Tuple, List, Union, Optional
import numpy as np

@dataclass(frozen=True)
class joint():
    joint_name: str
    joint_angle: np.float32
    joint_speed: np.float32
    joint_torque: np.float32
    time: float
    
@dataclass(frozen=True)
class camera():
    camera_name: str
    camera_fps: float
    camera_resolution: float
    camera_image = np.ndarray
    time: float

@dataclass(order=True, frozen=True)
class recorded_data():
    joint_data: List[joint]
    camera_data: List[camera]
    time_list: list 
    sort_index: float = field(init=False, repr=False)
    
    def __post_init__(self):
        self.sort_index = len(self.time_list)
        
@dataclass(order=True, frozen=True)
class processed_data(recorded_data):
    joint_data: List[joint]
    camera_data: List[camera]
    time_list: list 
    sort_index: float = field(init=False, repr=False)
    
    def __post_init__(self):
        self.sort_index = len(self.time_list) 
```
