## 7. Python函数中的参数类型
在 Python 函数中，参数的类型可以分为以下几种：

### 1. **位置参数 (Positional Arguments)**

位置参数是最常见的参数类型。它们是通过函数调用时根据位置传递的参数，顺序必须与函数定义中的参数顺序一致。

#### 示例：
```python
def greet(name, age):
    print(f"Hello {name}, you are {age} years old.")

greet("Alice", 30)  # 输出: Hello Alice, you are 30 years old.
```

在调用时，参数 `"Alice"` 和 `30` 分别对应 `name` 和 `age`，并按照顺序传递。

### 2. **默认参数 (Default Arguments)**

默认参数是函数定义时给定默认值的参数。如果调用时没有提供该参数的值，就会使用默认值。

#### 示例：
```python
def greet(name, age=25):
    print(f"Hello {name}, you are {age} years old.")

greet("Alice")    # 输出: Hello Alice, you are 25 years old.
greet("Bob", 30)  # 输出: Hello Bob, you are 30 years old.
```

在这个例子中，`age` 参数有默认值 `25`，如果没有提供 `age` 的值，默认使用 `25`。

### 3. **可变参数 (Variable-Length Arguments)**

可变参数允许函数接受任意数量的位置参数或关键字参数。

#### 3.1 **位置可变参数（*args）**

`*args` 允许你传递任意数量的位置参数，它会将这些参数作为元组 (`tuple`) 传递到函数内部。

#### 示例：
```python
def greet(*names):
    for name in names:
        print(f"Hello {name}!")

greet("Alice", "Bob", "Charlie")  # 输出: Hello Alice! Hello Bob! Hello Charlie!
```

在这个例子中，`*names` 接收所有传递给函数的位置参数，并将它们作为元组处理。

#### 3.2 **关键字可变参数（**kwargs）**

`**kwargs` 允许你传递任意数量的关键字参数，它会将这些参数作为字典 (`dict`) 传递到函数内部。

#### 示例：
```python
def greet(**person_info):
    for key, value in person_info.items():
        print(f"{key}: {value}")

greet(name="Alice", age=30, city="New York")
# 输出:
# name: Alice
# age: 30
# city: New York
```

在这个例子中，`**person_info` 接收所有传递给函数的关键字参数，并将它们作为字典处理。

### 4. **关键字参数 (Keyword Arguments)**

关键字参数是通过参数名来传递的参数，调用时可以不按顺序传递，只需要指定参数名。

#### 示例：
```python
def greet(name, age):
    print(f"Hello {name}, you are {age} years old.")

greet(name="Alice", age=30)  # 输出: Hello Alice, you are 30 years old.
```

在调用时，`name` 和 `age` 的顺序不重要，只要通过 `name="Alice"` 和 `age=30` 显式指定。

## 11. Python的深浅拷贝
[图示分析](https://www.runoob.com/w3cnote/python-understanding-dict-copy-shallow-or-deep.html)

````python
a = b = [1 , 2, [3, 4]]
a[2] = 3
print(a, b)
````

浅拷贝只成功独立拷贝了列表的外层，而列表的内层列表，还是共享的

**实际场景**： 状态变量（如关节角度、速度）被多个模块（控制器、规划器）同时读取。
如果采用浅拷贝，当一个模块修改状态时，其他模块也会感知到变化。

````python
import copy
list1 = [1, 2, 3, [1, 2]]
list2 = copy.copy(list1)
list2.append('a')
list2[3].append('a')
print(f'list1: {list1}')
print(f'list2: {list2}')
````

深拷贝使得两个列表完全独立开来，每一个列表的操作，都不会影响另一个

**实际场景**：在多线程或多进程中处理机器人数据（如路径规划），需要确保线程间的数据独立。

````python
import copy
list1 = [1, 2, 3, [1, 2]]
list3 = copy.deepcopy(list1)
list3.append('a')
list3[3].append('a')
print(f'list1: {list1}')
print(f'list2: {list2}')
````

### **Python闭包（Closure）**

闭包是**函数式编程**中的一个概念，在 Python 中指的是**内部函数**（嵌套函数）对外部函数作用域中**非全局变量的引用**，即使外部函数已经返回了，内部函数仍然能够访问这些变量。

简单来说，闭包是指：
- 一个函数被定义在另一个函数的内部。
- 内部函数引用了外部函数的变量。
- 外部函数执行完毕后，其作用域内的变量依然可以被内部函数访问。

---

### **闭包的条件**
一个闭包需要满足以下三个条件：
1. **必须有一个嵌套函数**（即函数定义在另一个函数内部）。
2. **内部函数引用了外部函数的变量**。
3. **外部函数返回了内部函数**，并且该内部函数能够在外部被使用。

---

### **闭包的原理**
当外部函数返回内部函数时，外部函数的局部作用域不会被销毁，因为内部函数依然引用了它。这样，外部函数的变量就会被保存在**一个特殊的环境（closure）**中。

---

### **闭包的代码示例**

#### 示例 1: 简单闭包
```python
def outer_function(x):
    def inner_function(y):
        return x + y
    return inner_function

closure = outer_function(10)  # 返回 inner_function
print(closure(5))  # 10 + 5 = 15
```

- **外部函数 `outer_function`** 定义了变量 `x`。
- **内部函数 `inner_function`** 使用了 `x`。
- 当 `outer_function` 执行完毕后，`x` 的值被捕获到 `inner_function` 的闭包中，因此 `closure` 能够记住 `x=10`。

#### 示例 2: 查看闭包的内容
我们可以通过函数对象的 `__closure__` 属性查看闭包保存的内容：
```python
def outer_function(x):
    def inner_function(y):
        return x + y
    return inner_function

closure = outer_function(10)
print(closure.__closure__)  # (<cell at ...: int object at ...>,)
print(closure.__closure__[0].cell_contents)  # 10
```

---

### **闭包的实际应用**
#### 1. **保存状态**
闭包可以用来创建具有记忆功能的函数，常见于计数器等应用。

```python
def counter():
    count = 0
    def increment():
        nonlocal count  # 修改外部变量需要用 nonlocal 声明
        count += 1
        return count
    return increment

counter_func = counter()
print(counter_func())  # 1
print(counter_func())  # 2
print(counter_func())  # 3
```

#### 2. **数据隐藏**
闭包可以实现某种程度的数据隐藏，将数据封装在外部函数中，仅通过内部函数访问。

```python
def make_multiplier(factor):
    def multiply(number):
        return factor * number
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(5))  # 10
print(triple(5))  # 15
```

#### 3. **装饰器**
闭包是实现装饰器的基础，装饰器本质上是一个返回闭包的函数。

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Function is being called...")
        result = func(*args, **kwargs)
        print("Function has finished.")
        return result
    return wrapper

@decorator
def say_hello(name):
    print(f"Hello, {name}!")

say_hello("Junhao")
```

---

### **闭包的注意事项**
1. **变量的延迟绑定问题**：
   - Python 的闭包中，内部函数会引用外部变量的**引用**，而不是直接保存值。
   - 如果外部变量在闭包执行时已经被修改，闭包中的值也会受影响。

   **示例问题**：
   ```python
   def create_multipliers():
       return [lambda x: i * x for i in range(4)]

   multipliers = create_multipliers()
   print([m(2) for m in multipliers])  # 输出 [6, 6, 6, 6] 而不是 [0, 2, 4, 6]
   ```

   **解决方法**：
   ```python
   def create_multipliers():
       return [lambda x, i=i: i * x for i in range(4)]

   multipliers = create_multipliers()
   print([m(2) for m in multipliers])  # 输出 [0, 2, 4, 6]
   ```

2. **变量作用域**：
   - 在闭包中修改外部函数变量时，需要使用 `nonlocal` 关键字。

   **示例**：
   ```python
   def outer():
       x = 10
       def inner():
           nonlocal x
           x += 1
           return x
       return inner

   func = outer()
   print(func())  # 11
   print(func())  # 12
   ```

## 28. 实现二分法查找函数

二分查找算法也称折半查找，基本思想就是折半，对比大小后再折半查找，必须是有序序列才可以使用二分查找。

````python
def binary_search(data, item):
    left = 0
    right = len(data) - 1
    
    while left < right:
        mid = (left + right) // 2
        if data[mid] == item:
            return True, mid
        elif data[mid] < item:
            left = mid + 1
        else:
            right = mid -1
    return False
````

````python
def binary_search_recursive(data, item):
    n = len(data)
    while n > 0:
        mid = n // 2
        if data[mid] == item:
            return True, mid
        elif data[mid] < item:
            return binary_search_recursive(data[mid:], item)
        elif data[mid] > item:
            return binary_search_recursive(data[:mid+1], item)
    
    return False
        
````

## 36. 单下划线和双下划线的作用

1. `__foo__`：一种约定，Python内部的名字，用来区别其他用户自定义的命名，以防冲突，就是例如`__init__()`, `__del__()`, `__call__()`等特殊方法
2. `_foo`：一种约定，用来指定变量私有。不能用`from model import *`导入，其他方便和公有变量一样访问
3. `__foo`：这个有真正的意义：解析器用`_classname__foo`来代替这个名字，以区别和其他类相同的命名，它无法直接像公有成员一样随便访问，通过`对象名._类名__xxx`这样的方式可以访问

````python
from dataclasses import dataclass

@dataclass
class people(object):
    ''' define a people with basic info'''
    name: str
    age: int
    __gender: str
    address: str = 'Nanchang'

john = people('john', 16, 'male')
print(john._people__gender)
````

````python
from typing import Literal, Optional
from functools import wraps

# 定义一个装饰器，用于标记接口方法
def abstract_method(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        raise NotImplementedError(f"The method {func.__name__} must be implemented in subclasses.")
    return wrapper

class Robot:
    """Basic robot class"""
    def __init__(self, name: str, type: Literal["humanoid", "arm", "legged", "wheel"], dof: Optional[int] = None):
        self.__name = name
        self.__dof = dof

        if type not in ["humanoid", "arm", "legged", "wheel"]:
            raise ValueError(f"Invalid robot type: {type}")
        self.__type = type

    @abstract_method
    def get_state(self):
        ...

    @abstract_method
    def get_observation(self):
        ...

    @abstract_method
    def action(self):
        ...

    @abstract_method
    def reset(self):
        ...

    @abstract_method
    def shutdown(self):
        ...

# 子类 1: Arm_Robot
class Arm_Robot(Robot):
    def __init__(self, name: str, dof: int):
        super().__init__(name=name, type="arm", dof=dof)

    def get_state(self):
        return f"Arm {self._Robot__name} is in a neutral position."

    def get_observation(self):
        return {"position": [0.0] * self._Robot__dof, "force": [0.0] * self._Robot__dof}

    def action(self):
        print(f"Moving arm {self._Robot__name} to target position.")

    def reset(self):
        print(f"Resetting arm {self._Robot__name} to default position.")

    def shutdown(self):
        print(f"Shutting down arm {self._Robot__name}.")

# 子类 2: Legged_Robot
class Legged_Robot(Robot):
    def __init__(self, name: str, dof: int):
        super().__init__(name=name, type="legged", dof=dof)

    def get_state(self):
        return f"Legged {self._Robot__name} is walking."

    def get_observation(self):
        return {"joint_angles": [45.0] * self._Robot__dof, "velocity": [1.0] * self._Robot__dof}

    def action(self):
        print(f"Legged {self._Robot__name} is running to the destination.")

    def reset(self):
        print(f"Resetting legged {self._Robot__name} to standing position.")

    def shutdown(self):
        print(f"Shutting down legged {self._Robot__name}.")

# 子类 3: Humanoid (多重继承)
class Humanoid(Arm_Robot, Legged_Robot):
    def __init__(self, name: str, dof: int):
        # 使用 super() 初始化，并手动传递所需参数
        Robot.__init__(self, name=name, type="humanoid", dof=dof)

    def get_state(self):
        arm_state = Arm_Robot.get_state(self)
        leg_state = Legged_Robot.get_state(self)
        return f"Humanoid {self._Robot__name} states: [Arm: {arm_state}, Legs: {leg_state}]"

    def get_observation(self):
        arm_obs = Arm_Robot.get_observation(self)
        leg_obs = Legged_Robot.get_observation(self)
        return {
            "arm": arm_obs,
            "legs": leg_obs,
        }

    def action(self):
        Arm_Robot.action(self)
        Legged_Robot.action(self)
        print(f"Humanoid {self._Robot__name} is performing a combined action.")

    def reset(self):
        Arm_Robot.reset(self)
        Legged_Robot.reset(self)
        print(f"Humanoid {self._Robot__name} fully reset.")

    def shutdown(self):
        Arm_Robot.shutdown(self)
        Legged_Robot.shutdown(self)
        print(f"Humanoid {self._Robot__name} fully shut down.")

# 子类 4: Wheel_Robot
class Wheel_Robot(Robot):
    def __init__(self, name: str):
        super().__init__(name=name, type="wheel", dof=4)

    def get_state(self):
        return f"Wheel robot {self._Robot__name} is rolling smoothly."

    def get_observation(self):
        return {"wheel_speeds": [10.0] * self._Robot__dof}

    def action(self):
        print(f"Wheel robot {self._Robot__name} is driving to the target location.")

    def reset(self):
        print(f"Resetting wheel robot {self._Robot__name}.")

    def shutdown(self):
        print(f"Shutting down wheel robot {self._Robot__name}.")

class RobotFactory:
    @staticmethod
    def create_robot(name: str, robot_type: str, dof: int = None):
        if robot_type == "arm":
            return Arm_Robot(name, dof)
        elif robot_type == "legged":
            return Legged_Robot(name, dof)
        elif robot_type == "humanoid":
            return Humanoid(name, dof)
        elif robot_type == "wheel":
            return Wheel_Robot(name)
        else:
            raise ValueError(f"Invalid robot type: {robot_type}") 
create a class for me

# 测试代码
if __name__ == "__main__":
    arm_robot = Arm_Robot("ArmBot", dof=7)
    legged_robot = Legged_Robot("LegBot", dof=6)
    humanoid_robot = Humanoid("HumanoidBot", dof = 30)
    wheel_robot = Wheel_Robot("WheelBot")

    for robot in [arm_robot, legged_robot, humanoid_robot, wheel_robot]:
        print(robot.get_state())
        robot.action()
        robot.reset()
        robot.shutdown()


````

## 40. 类中的各种函数
主要分为实例方法，类方法和静态方法

````python
from dataclasses import dataclass
import math as m

@dataclass
class Circle:
    radius: float
    
    # 静态方法：使用装饰器@staticmethod。参数随意，没有“self”和“cls”参数，但是方法体中不能使用类或实例的任何属性和方法。
    # 即在静态方法中，不会涉及到类中的方法和属性的操作。可以理解为将静态方法存在此类的名称空间中。
    # 调用：实例对象和类对象都可以调用。
    # 继承后子类可以直接调用
    @staticmethod
    def pi()-> float:
        return m.pi
    
    # 类方法: 使用装饰器@classmethod。第一个参数必须是当前类，参数名一般约定为cls, 通过它来传递类的属性和方法（不能传实例的属性和方法）。
    # 调用：实例对象和类对象都可以调用。
    # 继承后绑定在子类上，参数变为子类
    @classmethod
    def unit_circle(cls):
        return cls(radius=1.0)
    
    # 实例方法：该参数名一般约定为self, 通过它来传递实例的属性和方法（也可以传类的属性和方法）。
    # 调用：只能由实例对象调用。
    def area(self) -> float:
        return self.pi() * self.radius**2
````

## 43. 单例模式和工厂模式
- 单例模式：主要目的是确保某一个类只有一个实例存在；
- 工厂模式：包涵一个超类，这个超类提供一个抽象化的接口来创建一个特定类型的对象，而不是决定哪个对象可以被创建

**单例模式 (`singleton pattern`)**：

这种模式涉及到一个单一的类，该类负责创建自己的对象，同时确保只有单一对象被创建。这个类提供了一种访问其唯一对象的方式，可以直接访问，不需要实例化该类的对象
- 意图：保证一个类仅有一个实例，并提供一个访问它的全局访问点
- 主要解决：一个全局使用的类频繁地创建与销毁, 多个实例对共享资源的访问可能引发冲突。

````python
class Singleton:
    _instance = None  # 私有类变量，用于存储单例实例

    def __new__(cls, *args, **kwargs):
        if not cls._instance:  # 如果实例不存在，则创建
            cls._instance = object.__new__(cls)  # 直接调用 object.__new__(cls)
        return cls._instance


# 示例类
class FileHandler(Singleton):
    def __init__(self, filename):
        if not hasattr(self, 'initialized'):  # 确保 init 方法只执行一次
            self.filename = filename
            self.initialized = True

    def read(self):
        print(f"Reading from {self.filename}")

    def write(self, data):
        print(f"Writing {data} to {self.filename}")


# 测试
handler1 = FileHandler("file1.txt")
handler2 = FileHandler("file2.txt")

print(handler1 is handler2)  # True，证明两者是同一个实例
handler1.read()             # Reading from file1.txt
handler2.write("Hello")     # Writing Hello to file1.txt
````

````python
import threading

class Singleton:
    _instance = None
    _lock = threading.Lock()  # 锁对象，确保线程安全

    def __new__(cls, *args, **kwargs):
        if not cls._instance:  # 如果实例不存在
            with cls._lock:  # 加锁，确保只有一个线程能进入此代码块
                if not cls._instance:  # 双重检查，防止并发创建实例
                    cls._instance = object.__new__(cls)
        return cls._instance


class FileHandler(Singleton):
    _read_write_lock = threading.Lock()  # 读写操作的锁

    def __init__(self, filename):
        if not hasattr(self, 'initialized'):  # 确保 init 只执行一次
            self.filename = filename
            self.initialized = True

    def read(self):
        with self._read_write_lock:  # 确保读操作线程安全
            print(f"Reading from {self.filename}")

    def write(self, data):
        with self._read_write_lock:  # 确保写操作线程安全
            print(f"Writing {data} to {self.filename}")
            
def thread_function(name, handler):
    if name % 2 == 0:
        handler.read()
    else:
        handler.write(f"Data from thread {name}")

handler = FileHandler("file1.txt")

threads = []
for i in range(5):  # 创建多个线程
    t = threading.Thread(target=thread_function, args=(i, handler))
    threads.append(t)
    t.start()

for t in threads:
    t.join()
````

**工厂类模式 (`Factory Pattern`)**：

工厂类模式设计的核心是：让“生产”和“产品”解耦。
工厂模式的主要解决的问题：将原来分布在各个地方的对象创建过程单独抽离出来，交给工厂类负责创建。其他地方想要使用对象直接找工厂（即调用工厂的方法）获取对象。
优点：

- 一个调用者想创建一个对象，只要知道其名称就可以了。
- 扩展性高，如果想增加一个产品，只要扩展一个工厂类就可以。
- 屏蔽产品的具体实现，调用者只关心产品的接口。



                Product（产品抽象类/接口）
                   ↑
      +------------+-------------+
      |                          |
 ConcreteProductA         ConcreteProductB（具体产品类）
             ↑                       ↑
          SimpleFactory（工厂类）

````python
# 产品抽象类
class Product:
    def use(self):
        pass

# 具体产品类 A
class ConcreteProductA(Product):
    def use(self):
        print("Using Product A")

# 具体产品类 B
class ConcreteProductB(Product):
    def use(self):
        print("Using Product B")

# 工厂接口
class Factory:
    def create_product(self):
        pass

# 具体工厂类 A
class ConcreteFactoryA(Factory):
    def create_product(self):
        return ConcreteProductA()

# 具体工厂类 B
class ConcreteFactoryB(Factory):
    def create_product(self):
        return ConcreteProductB()

# 客户端代码
factory_a = ConcreteFactoryA()
product_a = factory_a.create_product()
product_a.use()  # Output: Using Product A

factory_b = ConcreteFactoryB()
product_b = factory_b.create_product()
product_b.use()  # Output: Using Product B
````

## 51. Python中的反射
反射就是通过字符串的形式，导入模块；通过字符串的形式，去模块寻找指定函数，并执行。利用字符串的形式去对象（模块）中操作（查找/获取/删除/添加）成员，一种基于字符串的事件驱动。简单理解就是用来判断某个字符串是什么，是变量还是方法。

````python
class NewClass(object):
    def __init__(self, name, male):
        self.name = name
        self.male = male
    
    def myname(self):
        print(f'My name is {self.name}')
        
    def mymale(self):
        print(f"I am a {self.male}")
        
# 创建一个 NewClass 实例
people = NewClass('luobo', 'boy')

# 使用 hasattr 判断 'name' 属性是否存在
print(hasattr(people, 'name'))

# 使用 getattr 获取 'name' 属性的值
print(getattr(people, 'name'))

# 使用 setattr 修改 'name' 属性的值为 'john'
setattr(people, 'name', 'john')
print(getattr(people, 'name'))

# 使用 delattr 删除 'name' 属性
delattr(people, 'name')

# 尝试访问被删除的属性 'name'
print(people.name)
````

````python
class Operations:
    def add(self, a, b):
        return a + b
    
    def subtract(self, a, b):
        return a - b

# 配置文件
config = {
    "operation": "add",
    "args": [10, 5]
}

# 动态执行配置中定义的操作
operations = Operations()
operation_name = config["operation"]

if hasattr(operations, operation_name):
    operation = getattr(operations, operation_name)
    result = operation(*config["args"])  # 动态传参并执行
    print(f"Result: {result}")
````

- `key`: 一个只有一个参数的函数，这个函数会被用在序列里的每一个元素上，所产生的结果将是排序算法依赖的对比关键字。比如说，在对一些字符串排序时，在对一些字符串排序时，可以用`key=str.lower`来实现忽略大小写的排序，或者是用`key=len`进行基于字符串长度的排序。这个参数的默认值是恒等函数，也就是默认用元素自己的值来排序。

## 56 字符串类型
Python supports several types of string literals, each serving different purposes. Here's a breakdown of **f-strings**, **r-strings**, and **b-strings**:

| Feature       | **f-string**                  | **r-string**              | **b-string**              |
|---------------|-------------------------------|---------------------------|---------------------------|
| **Prefix**    | `f` or `F`                    | `r` or `R`                | `b` or `B`                |
| **Purpose**   | Embed expressions in strings  | Ignore escape sequences   | Represent byte data       |
| **Example**   | `f"My name is {name}"`        | `r"C:\path\file"`         | `b"hello"`                |

`f-strings` - Used for embedding expressions inside string literals.

````python
name = "Junhao"
age = 25
f_string = f"My name is {name} and I am {age} years old."
print(f_string)  # Output: My name is Junhao and I am 25 years old.
````

`r-strings` - Used for strings where escape sequences (like \n, \t) should not be interpreted.

````python
raw_string = r"C:\Users\Junhao\Documents"
print(raw_string)  # Output: C:\Users\Junhao\Documents
````

````python
# Without r, you'd need to escape backslashes:
regular_string = "C:\\Users\\Junhao\\Documents"
print(regular_string)  # Output: C:\Users\Junhao\Documents
````

`b-strings` - Used for representing byte data, especially in file I/O or network communication.

````python
text = "Hello, World!"
byte_string = text.encode("utf-8")  # Convert to bytes
print(byte_string)  # Output: b'Hello, World!'

decoded_string = byte_string.decode("utf-8")  # Convert back to string
print(decoded_string)  # Output: Hello, World!
````

## 59. 函数式编程
函数式编程是一种抽象程度很高的编程范式，纯粹的函数式编程语言编写的函数没有变量，因此，任意一个函数，只要输入是确定的，输出就是确定的，这种纯函数称之为没有副作用。而允许使用变量的程序设计语言，由于函数内部的变量状态不确定，同样的输入，可能得到不同的输出，因此，这种函数是有副作用的。由于 Python 允许使用变量，因此，Python 不是纯函数式编程语言。

函数式编程的一个特点就是，允许把**函数本身作为参数传入另一个函数，还允许返回一个函数**

````python
def sum(*args):
    def inner_sum():
        tmp = 0
        for i in args:
            tmp += i
        return tmp
    return inner_sum # 函数作为返回值

mysum = sum(2, 4, 6)
print(type(mysum))
mysum()
````

## 60. 简述闭包
如果在一个内部函数里，对在外部作用域（但不是在全局作用域）的变量进行引用，那么内部函数就被认为是闭包（closure）。

## 65. 迭代器的使用

根据维基百科，<u>迭代器是⼀个让程序员可以遍历⼀个容器的对象。然⽽，⼀个迭代器在遍历并读取⼀个容器的数据元素时，并不会执⾏⼀个迭代</u>。简单来说这里有三部分：
- 可迭代对象（Iterable）: Python中任意的对象，只要它定义了可以返回一个迭代器的`__iter__`方法或者定义了可以支持下标索引的`__getitem__`方法，那么它就是一个可迭代对象。简单说，**可迭代对象就是提供迭代器的任意对象**。
- 迭代器（Iterator）:任意对象，只要定义了`__next__`方法，它就是一个迭代器。
- 迭代 （Iteration）:简单来讲，它就是从某个地方取出一个元素的过程。当我们使用循环来遍历某个东西时，这个过程本身就叫迭代。

迭代器是一个可以记住遍历位置的对象。迭代器对象从集合的第一个元素开始访问，直到所有的元素被访问完结束。**迭代器只能往前不会后退**。

---

Python中迭代器协议主要用到了两个魔法方法：`__iter__()`, `__next__()`
- `__iter__()`方法创建一个具有`__next__()`方法的迭代器对象
- `__next__()`方法返回下一个迭代器对象



````python
# 字符串，列表或元组对象都可用于创建迭代器：
list = [1, 2, 3, 4]
it = iter(list)
print(next(it))
print(next(it))

# 迭代器对象可以使用常规for语句进行遍历：
list = [1, 2, 3, 4]
it = iter(list)
for x in it:
    print(x, end = ' ')
````

### 创建迭代器

把一个类作为一个迭代器使用需要在类中实现两个方法：`__iter__()`与`__next__()`。
- `__iter__()`方法返回一个特殊的迭代器对象，这个迭代器对象实现了`__next__()`方法并通过`StopIteration`异常标识迭代的完成。
- `__next__()`方法会返回下一个迭代器对象。

````python
class MyNumber:
    def __iter__(self):
        self.a = 1
        return self
    
    def __next__(self):
        x = self.a
        self.a += 1
        return x
    
myclass = MyNumber()
myiter = iter(myclass)
for x in myiter:
    print(x)
````

### StopIteration
`StopIteration`异常用于标识迭代的完成，防止出现无限循环的情况，在`__next__()`方法中我们可以设置在完成指定循环次数后触发`StopIteration`异常来结束迭代。


````python
class MyNumber:
    def __iter__(self):
        self.a = 1
        return self
        
    def __next__(self):
        if self.a < 20:
            x = self.a 
            self.a += 1
            return x
        else:
            raise StopIteration
        
num = MyNumber()
myiter = iter(num)
for i in myiter:
    print(i, end=' ')
````

## 66. 生成器的使用

生成器（generator）是Python中一种特殊的迭代器，能够通过边循环边计算的方式有效处理大量数据，从而减少内存消耗和提升性能。以下是生成器的核心特点和优势的总结：
1. 边循环边计算
    - 生成器不会一次性将所有值存储在内存中，而是在需要时动态生成值。
    - 通过`yield`关键字逐步生成数据。
2. 迭代器特性
    - 生成器是迭代器，支持 `next()` 和迭代操作，但只能被迭代一次。
    - 每次遇到 yield 时会暂停函数执行，保存当前状态，并返回 yield 的值。
    - 下一次迭代会从上次暂停的地方继续执行。
3. 延迟计算
    - 生成器延迟计算数据，避免不必要的计算，特别适合处理大规模或无限序列的数据。
4. 实现简单
    - 生成器函数使用普通函数的语法，只需在适当位置使用`yield`替代`return`。

### 实现生成器的两种方式

**第一种方法**：把一个列表生成式的[]改成()，就创建了一个生成器。

````python

L = [x**2 for x in range(10)]
print(L)

g = (x**2 for x in range(10))
print(g)
````

**第二种方法**：在函数中使用yield关键字，函数就变成了一个生成器。调用该函数，就等于创建了一个生成器对象。
函数里有了yield后，执行到yield就会停住，当需要再往下算时才会再往下算。所以生成器函数即使是有无限循环也没关系，它需要算到多少就会算多少，不需要就不往下算。一个生成器，主要是通过循环反复调用next()方法，直到捕获异常。`for`循环会自动捕捉到这个异常并停止调用`next()`


````python
# 使用yield之后, 此类函数就称为生成器函数
# 调用生成器函数，就可以创建一个num生成器对象
def intNum():
    print("开始执行")
    for i in range(5):
        yield i
        print("继续执行")
num = intNum()
````

**生成器的使用有三种方法**：

因此，带yield的函数具体内部执行操作为：
- yield方法：相当于return作用，程序遇到yield则直接中止后续步骤；
- 当再次调用生成器时，next()方法会唤醒，并继续执行yield后续步骤；
- 还可以调用send()方法，可以唤醒，并传入一个值，继续执行yield后续步骤。

生成器是可迭代的，每一次只可读一次。因此常常与for循环一起组合使用。

````python
#调用 next() 内置函数
print(next(num))

#调用 __next__() 方法
print(num.__next__())

#通过for循环遍历生成器
for i in num: print(i)
````

3. `send()`方法

**注意**：如果生成器完成（运行到 `return` 语句）后再次调用 `send` 或 `next`，会抛出 `StopIteration` 异常，其值是 `return` 语句的返回值（这里是 '没有更多元素了'）。

````python
# 1. 与简单的 next() 方法相比，send(value) 不仅会让生成器运行到下一个 yield，还会将 value 作为当前 yield 表达式的返回值。
# 2. 这使得生成器能够根据外部输入动态调整其行为。

def gen():
    i = 0
    while i < 5:
        temp = yield i
        print('temp: ', temp)
        i += 1
    return '没有更多元素了'

g = gen()
print(g.send(None))

n1 = g.send('起点')
print('n1', n1)

n2 = g.send('发展')
print('n2', n2)
````

动态调整生成器逻辑

````python
def counter():
    count = 0
    while True:
        increment = yield count  # 接收 send() 的值作为 increment
        if increment is None:    # 如果没有通过 send() 传递值，默认增量为 1
            increment = 1
        count += increment

gen = counter()
print(next(gen))       # 输出 0，启动生成器
print(gen.send(5))     # 传入增量 5，count 增加到 5，输出 5
print(gen.send(2))     # 传入增量 2，count 增加到 7，输出 7
print(gen.send(None))  # 未传入值，默认增量为 1，count 增加到 8，输出 8
````

````python
def echo():
    while True:
        received = yield  # 接收外部输入
        print(f"Received: {received}")

gen = echo()
next(gen)            # 启动生成器（运行到第一个 yield 暂停）
gen.send("Hello")    # 传递数据，生成器内部打印 "Received: Hello"
gen.send("World")    # 生成器内部打印 "Received: World"
````

**可迭代对象(Iterable)** 是可以直接作用于for循环的对象。包括集合数据类型（list、tuple、dict、set、str等）和生成器（generator）。可以使用isinstance()判断一个对象是否是Iterable对象。

````python
from collections.abc import Iterable, Iterator
isinstance([], Iterable)
isinstance({}, Iterable)
isinstance('abc', Iterable)
isinstance((x for x in range(10)), Iterable)
isinstance(100, Iterable)
````

**迭代器（Iterator）** 表示的是一个数据流。`Iterator`对象可以被`next()`函数调用并不断返回下一个数据，直到没有数据时抛出`StopIteration`错误。可以把这个数据流看做是一个有序序列，但我们却不能提前知道序列的长度，只能不断通过`next()`函数实现按需计算下一个数据，所以`Iterator`的计算是惰性的，只有在需要返回下一个数据时它才会计算。`Iterator`甚至可以表示一个无限大的数据流，例如全体自然数。而使用`list`是永远不可能存储全体自然数的。

**生成器都是Iterator对象**，但list、dict、str虽然是Iterable，却不是Iterator。把list、dict、str等Iterable变成Iterator可以使用iter()函数：

````python
isinstance(iter([]), Iterator)
isinstance(iter('abc'), Iterator)
````

## 67. 冒泡排序

````python
from typing import Sequence
import random

def bubble_sort(data: Sequence[int], if_reversed: bool) -> Sequence[int]:
    for i in range(len(data) - 1):
        for j in range(len(data) - i - 1):
            if data[j] > data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]
    
    if if_reversed:
        data.reverse()
        
    return data

L1 = [random.randint(1,15) for _ in range(10)]
print(bubble_sort(L1, False))
print(bubble_sort(L1, True))
````

## 69. 使用两种方法实现`switch`

1. 使用`match-case`, Python3.10引入了`match`, 用来实现类似`switch`的语法。

````python
def switch(case_value):
    match case_value:
        case 1:
            return "Case 1"
        case 2:
            return "Case 2"
        case 3:
            return "Case 3"
        case _:
            return "Default case"

print(switch(1))  # 输出: Case 1
print(switch(99))  # 输出: Default case     
````

## 71. 比较两个json数据是否相等

1. **方法一**
- json数据转换成字典；
- 将两个字典按`key`排好序，然后使用`zip()`函数将两个字典对应的元素打包成元组。比较对应的元素的`value`是否相等
- `zip()`函数用于将可迭代的对象作为参数，将对象中对应的元素打包成一个个元组。

````python
import json

file_text1 = '{"name":"john","age":22,"sex":"woman","address":"USA"}'
file_text2 = '{"name":"john","age":22,"sex":"man","address":"USA"}'

dict1 = json.loads(file_text1)
dict2 = json.loads(file_text2)
for i, key1, key2 in enumerate(zip(sorted(dict1), sorted(dict2))):
    if str(dict1[key1]) != str(dict2[key2]):
        raise ValueError("The two json file are not equal!")
````

2. **方法二**
- json数据转换成字典；
- 将两个字典按`key`排好序，然后先比较`key`是否一致
- 然后用`key`遍历字典，看对应的`value`是否一致

````python
import json

file_text1 = '{"name":"john","age":22,"sex":"woman","address":"USA"}'
file_text2 = '{"name":"john","age":22,"sex":"man","address":"USA"}'

# Assuming file_text1 and file_text2 are valid JSON strings
dict1 = json.loads(file_text1)
dict2 = json.loads(file_text2)

# Ensure both have the same keys
if sorted(dict1.keys()) != sorted(dict2.keys()):
    raise ValueError("The two JSON files have different keys!")

# Compare the values for each key
for i, key in enumerate(sorted(dict1.keys())):
    value1 = dict1[key]
    value2 = dict2[key]
    if value1 != value2:
        raise ValueError(f"Difference found at key '{key}' (Index: {i}): {value1} != {value2}")

print("The two JSON files are equal!")
````

## 85. 什么是异步非阻塞

同步异步指的是调用者与被调用者之间的关系。

- 所谓同步，就是在发出一个功能调用时，在没有得到结果之前，该调用就不会返回，一旦调用返回，就得到了返回值；
- 异步的概念和同步相对，调用在发出之后，这个调用就直接返回了，所以没有返回结果。当该异步功能完成后，被调用者可以通过状态、通知或回调来通知调用者。

阻塞非阻塞是线程或进程之间的关系。

- 阻塞调用是指调用结果返回之前，当前线程会被挂起（如遇到io操作）。调用线程只有在得到结果之后才会返回。函数只有在得到结果之后才会将阻塞的线程激活
- 非阻塞和阻塞的概念相对应，非阻塞调用指在不能立刻得到结果之前也会立刻返回，同时该函数不会阻塞当前线程

## 87. 退出Python后，内存是否全部释放

在 Python 程序退出后，内存是否全部释放的问题上，答案是否定的。以下是详细解释：

1. **对象循环引用**：Python 使用引用计数和垃圾回收机制来管理内存。如果两个或多个对象相互引用，形成循环引用，引用计数不会降为零，导致这些对象不会被自动回收。

2. **全局命名空间引用**：全局变量在程序退出时可能仍然存在引用，导致这些变量占用的内存不会被释放。

3. **C 库保留的内容**：如果 Python 程序使用了 C 扩展库，这些库可能会保留一些内存，即使 Python 解释器已经退出。这是因为 C 库的内存管理与 Python 的垃圾回收机制不同。

总结来说，虽然 Python 会尝试在退出时释放大部分内存，但由于上述原因，某些内存可能不会被完全释放。

````python
# 对象之间的循环引用
class A:
    def __init__(self):
        self.b = None

class B:
    def __init__(self):
        self.a = None

# 创建实例
a = A()
b = B()

# 相互引用
a.b = b
b.a = a

# 删除引用
del a
del b

# 即使删除了引用，a 和 b 仍然存在于内存中，因为它们相互引用
````

````python
# 全局命名空间引用
gloabl_var = []

def add_to_global(item):
    global gloabl_var
    gloabl_var.append(item)
    
add_to_global(1)
add_to_global(2)

print(gloabl_var)

# 这种情况下，global_var 在程序退出前不会被释放，因为它存在于全局命名空间中。
````

## 93.给出多进程和多线程的示例

### 多进程示例
多进程允许你创建独立的进程，每个进程都有自己的 Python 解释器和内存空间。这对于 CPU 密集型任务非常有用。

````python
import multiprocessing
import time

def worker(num):
    """进程工作函数"""
    print(f'进程工作: {num}')
    time.sleep(2)
    return num * num

if __name__ == '__main__':
    with multiprocessing.Pool(processes=4) as pool:
        results = pool.map(worker, range(10))
    print(results)
````

### 多线程示例
多线程允许你在同一个进程中创建多个线程。这对于 I/O 密集型任务非常有用。

````python
import threading
import time

def worker(num):
    """线程工作函数"""
    print(f'线程工作: {num}')
    time.sleep(2)
    print(f'线程工作 {num} 完成')

threads = []
for i in range(10):
    t = threading.Thread(target=worker, args=(i,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()
````

### 结合多进程和多线程
你可以结合多进程和多线程来高效处理 CPU 密集型和 I/O 密集型任务。以下是一个示例：

````python
import multiprocessing
import threading
import time

def io_bound_task(num):
    """I/O 密集型任务"""
    print(f'I/O 密集型任务: {num}')
    time.sleep(2)
    print(f'I/O 密集型任务 {num} 完成')

def cpu_bound_task(num):
    """CPU 密集型任务"""
    print(f'CPU 密集型任务: {num}')
    result = 0
    for i in range(1000000):
        result += i * i
    print(f'CPU 密集型任务 {num} 完成')
    return result

def worker(num):
    """同时运行 I/O 密集型和 CPU 密集型任务的工作函数"""
    io_thread = threading.Thread(target=io_bound_task, args=(num,))
    io_thread.start()
    io_thread.join()
    return cpu_bound_task(num)

if __name__ == '__main__':
    with multiprocessing.Pool(processes=4) as pool:
        results = pool.map(worker, range(10))
    print(results)
````

### 94. 如何使用装饰器来缓存函数的结果，以提高函数的性能。

使用 `lru_cache` 可以显著提高那些计算开销大且结果不经常变化的函数的性能。例如：

- 复杂的数学计算
- 需要频繁访问的数据库查询结果
- 需要频繁读取的文件内容

````python
from functools import lru_cache

# 意味着 expensive_computation 函数的缓存最多可以存储 128 个不同的输入参数及其对应的计算结果。
# 如果超过这个数量，最久未使用的缓存条目将被移除。
@lru_cache(maxsize=128) 
def expensive_computation(n):
    print(f"Computing {n}...")
    result = 0
    for i in range(n):
        result += i * i
    return result

# 第一次调用会计算结果并缓存
print(expensive_computation(1000))  # 输出: Computing 1000... 以及计算结果

# 第二次调用会直接从缓存中获取结果
print(expensive_computation(1000))  # 输出: 直接输出计算结果，没有 "Computing 1000..." 消息
````

在 Python 3.9 及更高版本中，`functools` 模块引入了一个新的装饰器 `@cache`，它是 `@lru_cache` 的简化版本，等效于 `@lru_cache(maxsize=None)`。这意味着它会缓存所有的调用结果，而不会限制缓存的大小。

````python
from functools import cache

# 意味着 expensive_computation 函数的缓存没有大小限制。
# 所有不同的输入参数及其对应的计算结果都会被缓存。
@cache
def expensive_computation(n):
    print(f"Computing {n}...")
    result = 0
    for i in range(n):
        result += i * i
    return result

# 第一次调用会计算结果并缓存
print(expensive_computation(1000))  # 输出: Computing 1000... 以及计算结果

# 第二次调用会直接从缓存中获取结果
print(expensive_computation(1000))  # 输出: 直接输出计算结果，没有 "Computing 1000..." 消息
````

### 95. 如何使用`@depreicated`来标记弃用

1. 弃用函数

````python
from deprecated import deprecated

@deprecated(version='1.2.0', reason="This function will be removed in version 2.0.0. Use new_function instead.")
def old_function(x, y):
    return x + y

def new_function(x, y):
    return x * y

# 使用示例
print(old_function(3, 4))  # 输出: 7，并发出弃用警告
print(new_function(3, 4))  # 输出: 12
````

2. 弃用类

````python
from deprecated import deprecated

@deprecated(reason="This class will be removed in future versions. Use NewClass instead.")
class OldClass:
    def method(self):
        return "Old method"

class NewClass:
    def method(self):
        return "New method"

# 使用示例
old_instance = OldClass()  # 创建实例时发出弃用警告
print(old_instance.method())  # 输出: Old method
new_instance = NewClass()
print(new_instance.method())  # 输出: New method
````
